```python
from collections import Counter
import string

text = """Python is easy to learn. Python is useful for data analysis.
Python is also useful for machine learning."""

clean_text = text.lower()

clean_text = clean_text.translate(
    str.maketrans('', '', string.punctuation)
)

words = clean_text.split()

word_count = len(words)

character_count = len(
    clean_text.replace(" ", "").replace("\n", "")
)

vowels = "aeiou"
vowel_count = sum(
    1 for char in clean_text if char in vowels
)

word_frequency = Counter(words)

repeated_words = {
    word: count
    for word, count in word_frequency.items()
    if count > 1
}

print("========== TEXT ANALYSIS ==========")
print("Original Text:")
print(text)

print("\nTotal Number of Words:", word_count)
print("Total Number of Characters:", character_count)
print("Total Number of Vowels:", vowel_count)

print("\nRepeated Words:")

for word, count in repeated_words.items():
    print(word, "->", count, "times")
```
